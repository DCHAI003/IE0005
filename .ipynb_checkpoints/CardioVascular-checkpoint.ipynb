```python
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
sb.set()

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, precision_recall_fscore_support

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
cardiodata = pd.read_csv('cardio_train.csv', sep=';')
cardiodata.head()
```

```python
# 1.inspect and clean cardiovascular dataset
print("Original shape:", cardiodata.shape)
print(cardiodata.columns.tolist())
cardiodata.head()
```

```python
cardiodata.info()
print(cardiodata.describe(include='all').T)

# Check missing values
print("Missing values per column:\n", cardiodata.isnull().sum())
```

```python
#Remove rows where systolic <= diastolic (ap_hi <= ap_lo)
before = cardiodata.shape[0]
cardiodata = cardiodata[cardiodata['ap_hi'] > cardiodata['ap_lo']]
print(f"Removed {before - cardiodata.shape[0]} rows with ap_hi <= ap_lo")

#Remove extreme heights and weights
cardiodata = cardiodata[(cardiodata['height'] >= 100) & (cardiodata['height'] <= 240)]
cardiodata = cardiodata[(cardiodata['weight'] >= 30) & (cardiodata['weight'] <= 250)]
print("After removing implausible height/weight rows:", cardiodata.shape)

#Drop any duplicates
before = cardiodata.shape[0]
cardiodata = cardiodata.drop_duplicates()
print("Removed duplicates:", before - cardiodata.shape[0])

#Drop all NA rows
before = cardiodata.shape[0]
cardiodata = cardiodata.dropna()
print("Removed NA rows:", before - cardiodata.shape[0])
```

```python
#2. Convert age to years if value>150
if cardiodata['age'].max() > 150:
    cardiodata['age_years'] = (cardiodata['age'] / 365).round().astype(int)
else:
    # if value<=150
    cardiodata['age_years'] = cardiodata['age'].astype(int)
    
print("Age (years) - min, max:", cardiodata['age_years'].min(), cardiodata['age_years'].max())
```

```python
#3. Compute BMI
# BMI = weight (kg) / (height (m))^2
cardiodata['height_m'] = cardiodata['height'] / 100.0
cardiodata['BMI'] = cardiodata['weight'] / (cardiodata['height_m'] ** 2)

print(cardiodata['BMI'].describe())
```

```python
# 4)Engineer risk features
# BMI risk: 1 if BMI >= 30 (obese), else 0 
cardiodata['bmi_risk'] = (cardiodata['BMI'] >= 30).astype(int)

# High blood pressure: systolic >= 140 OR diastolic >= 90 (1 = high BP)
cardiodata['high_bp'] = ((cardiodata['ap_hi'] >= 140) | (cardiodata['ap_lo'] >= 90)).astype(int)

# Cholesterol risk: dataset usually codes 1=normal, 2=above norm, 3=well above
# Mark 2 or 3 as risk
def cholesterol_risk(x):
    if x > 1:  
        return 1
    else:
        return 0
cardiodata['chol_risk'] = cardiodata['cholesterol'].apply(cholesterol_risk)

# Glucose risk: same as cholesterol risk
def glucose_risk(x):
    if x > 1:  
        return 1
    else:
        return 0
cardiodata['gluc_risk'] = cardiodata['gluc'].apply(glucose_risk)

# Lifestyle binary flags (smoke, alco, active) typically already 0/1 in dataset
# We define: smoke_risk: 1, alco_risk: 1, inactive_risk: 1 (active==1 => no risk)
cardiodata['smoke_risk'] = cardiodata['smoke'].astype(int)
cardiodata['alco_risk'] = cardiodata['alco'].astype(int)
cardiodata['inactive_risk'] = (cardiodata['active'] == 0).astype(int)

# check how many people are in each risk category
risk_cols = ['bmi_risk','high_bp','chol_risk','gluc_risk','smoke_risk','alco_risk','inactive_risk']
print(cardiodata[risk_cols].sum())
```

```python
# 5)Calculate Lifestyle Risk Index (LRI) on 0-8 scale.The higher the value (0–8), the higher the overall risk.
# cardiodata['LRI'] = (
#     cardiodata['bmi_risk'] +
#     cardiodata['high_bp'] +
#     cardiodata['cholesterol'] +  # already encoded as 1 or 2 or 3 for elevated levels
#     cardiodata['gluc']           # already encoded as 1 or 2 or 3 for elevated levels
# )
cardiodata['LRI'] = cardiodata[risk_cols].sum(axis=1).astype(int)
print("Lifestyle Risk Index (LRI) created. Summary:")
print(cardiodata['LRI'].describe())
```

```python
# 6) Save cleaned dataset for modelling
cardiodata.to_csv('cardio_cleaned.csv', index=False)
print(" Cleaned dataset saved as 'cardio_cleaned.csv'")
```

```python
cardio_cleaned = pd.read_csv('cardio_cleaned.csv')
print(cardio_cleaned.shape)
cardio_cleaned.head()
```

```python
# Columns that form LRI
RISK_PARTS = ['bmi_risk','high_bp','chol_risk','gluc_risk','smoke_risk','alco_risk','inactive_risk']

#Splitting baseline(without LRI) and enhanced(with LRI)
cols = cardio_cleaned.columns.tolist()
feature_cols_baseline = []
for c in cols:
    if c not in ['cardio', 'LRI']:
        feature_cols_baseline.append(c)
        
feature_cols_enhanced = []
for c in cols:
    if c not in ['cardio']+ RISK_PARTS:
        feature_cols_enhanced.append(c)

BASE_COLS = feature_cols_baseline.copy()
ENH_COLS  = feature_cols_enhanced.copy()

print(feature_cols_baseline) #without LRI
print(feature_cols_enhanced) #with LRI
print("Enhanced uses LRI:", 'LRI' in cardio_cleaned.columns)
overlap = set(feature_cols_baseline).intersection(feature_cols_enhanced)
print("Overlap count:", len(overlap))
```

```python
#train/test split for Baseline and enhanced so its fair
idx_train, idx_test = train_test_split(
    cardio_cleaned.index,
    test_size=0.2,
    random_state=42, #random_state is used for reproducibility
    stratify=cardio_cleaned['cardio'] #while stratify is to prevent inbalance
)

#BASELINE training set
Xb_train = cardio_cleaned.loc[idx_train, feature_cols_baseline]
Xb_test  = cardio_cleaned.loc[idx_test,  feature_cols_baseline]
y_train  = cardio_cleaned.loc[idx_train, 'cardio']
y_test   = cardio_cleaned.loc[idx_test,  'cardio']

print("Baseline Train Set :", Xb_train.shape, y_train.shape)
print("Baseline Test Set  :", Xb_test.shape, y_test.shape)

#Enhanced training set
Xe_train = cardio_cleaned.loc[idx_train, feature_cols_enhanced]
Xe_test  = cardio_cleaned.loc[idx_test,  feature_cols_enhanced]

print("Enhanced Train Set :", Xe_train.shape, y_train.shape)
print("Enhanced Test Set  :", Xe_test.shape, y_test.shape)
```

```python
Xb_train = Xb_train.reindex(columns=BASE_COLS)
Xb_test  = Xb_test.reindex(columns=BASE_COLS)

Xe_train = Xe_train.reindex(columns=ENH_COLS)
Xe_test  = Xe_test.reindex(columns=ENH_COLS)
```

```python
#function for finding threshold and then evalute
def eval_model(X_train, y_train, X_test, y_test, label,
                     max_depth=6, class_weight=None,
                     metric="accuracy", thres_grid=np.linspace(0.2, 0.8, 61)):
    
    #split train so that wont touch test
    idx_train2, idx_value = train_test_split(
        X_train.index,
        test_size=0.25,
        random_state=42,
        stratify=y_train
    )
    X_train2, y_train2 = X_train.loc[idx_train2], y_train.loc[idx_train2]
    X_value, y_value = X_train.loc[idx_value], y_train.loc[idx_val]

    #decision tree classifier
    DT = DecisionTreeClassifier(max_depth=max_depth, class_weight=class_weight, random_state=42)
    DT.fit(X_train2, y_train2)
    p_value = DT.predict_proba(X_value)[:,1]

    #try threshold from 0.2 to 0.8 -> convert each t to labels and get accuracy
    scores = [accuracy_score(y_value, (p_value >= t).astype(int)) for t in thres_grid]
    best_thres = float(thres_grid[int(np.argmax(scores))])

    #evaluate on test once with chosen thres
    p_test = DT.predict_proba(X_test)[:,1]
    yhat_test = (p_test >= best_t).astype(int)
    prec, rec, f1, _ = precision_recall_fscore_support(y_test, yhat_test, average="binary")

    row = {
        "model": label,
        "threshold": best_t,
        "accuracy": accuracy_score(y_test, yhat_test),
        "precision": prec,
        "recall": rec,
        "f1": f1,
        "roc_auc": roc_auc_score(y_test, p_test)  # AUC doesn't need threshold
    }
    return row, DT, yhat_test
```

```python
row_base, dt_base, yhat_base = eval_model(
    Xb_train, y_train, Xb_test, y_test,
    label="DecisionTree baseline (best thr)",
    max_depth=6, class_weight=None, metric="accuracy"
)

results = [row_base]

row_enhance, dt_enhance, yhat_enhance = eval_model(
        Xe_train, y_train, Xe_test, y_test,
        label="DecisionTree +LRI (best thr)",
        max_depth=6, class_weight=None, metric="accuracy"
)
results.append(row_enhance)
```

```python

performance = pd.DataFrame(results)
display(performance)

#printing out confusion matrix
plt.figure(figsize=(4.8, 4.2))
sb.heatmap(confusion_matrix(y_test, yhat_base),annot =True,fmt=".2f", cmap="Blues", cbar=False,
                xticklabels=["No CVD","CVD"], yticklabels=["No CVD","CVD"])
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("DecisionTree baseline — Confusion Matrix")
plt.tight_layout(); plt.show()


plt.figure(figsize=(4.8, 4.2))
sb.heatmap(confusion_matrix(y_test, yhat_enhance),annot =True,fmt=".2f", cmap="Blues", cbar=False,
                xticklabels=["No CVD","CVD"], yticklabels=["No CVD","CVD"])
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("DecisionTree +LRI — Confusion Matrix")
plt.tight_layout(); plt.show()
```

```python
#trying logistic regression+scaling

#Baseline
pipe_base = Pipeline([('scaler', StandardScaler(with_mean=False)), ('lr', LogisticRegression(max_iter=2000))])
pipe_base.fit(Xb_train, y_train)
acc_base_lr = accuracy_score(y_test, pipe_base.predict(Xb_test))

#Enhanced
pipe_enhanced  = Pipeline([('scaler', StandardScaler(with_mean=False)), ('lr', LogisticRegression(max_iter=2000))])
pipe_enhanced.fit(Xe_train, y_train)
acc_enhanced_lr = accuracy_score(y_test, pipe_enhanced.predict(Xe_test))

print("LR baseline accuracy:", acc_base_lr)
print("LR +LRI accuracy    :", acc_enhanced_lr)
```

```python
def scores(model, X, y, thr=0.5, name=""):
    p = model.predict_proba(X)[:,1]
    yhat = (p >= thr).astype(int)
    prec, rec, f1, _ = precision_recall_fscore_support(y, yhat, average="binary")
    return {
        "model": name, "threshold": thr,
        "accuracy": accuracy_score(y, yhat),
        "precision": prec, "recall": rec, "f1": f1,
        "roc_auc": roc_auc_score(y, p)
    }
```

```python
#Split train into train/val quickly (same idx_train set)
idx_tr2, idx_val = train_test_split(idx_train, test_size=0.25, random_state=42, stratify=cardio_cleaned.loc[idx_train, 'cardio'])

#Baseline val data
Xb_tr2 = cardio_cleaned.loc[idx_tr2, BASE_COLS].reindex(columns=BASE_COLS)
yb_tr2 = cardio_cleaned.loc[idx_tr2, 'cardio']
Xb_val = cardio_cleaned.loc[idx_val,  BASE_COLS].reindex(columns=BASE_COLS)
yb_val = cardio_cleaned.loc[idx_val,  'cardio']

#Enhanced val data
Xe_tr2 = cardio_cleaned.loc[idx_tr2, ENH_COLS].reindex(columns=ENH_COLS)
ye_tr2 = cardio_cleaned.loc[idx_tr2, 'cardio']
Xe_val = cardio_cleaned.loc[idx_val,  ENH_COLS].reindex(columns=ENH_COLS)
ye_val = cardio_cleaned.loc[idx_val,  'cardio']

pipe_base.fit(Xb_tr2, yb_tr2)
pipe_enhanced.fit(Xe_tr2, ye_tr2)

#best threshold
ths = np.linspace(0.2, 0.8, 61)

p_val_base = pipe_base.predict_proba(Xb_val)[:, 1]
accs_base = [accuracy_score(yb_val, (p_val_base >= t).astype(int)) for t in ths]
best_t_base = float(ths[int(np.argmax(accs_base))])

p_val_enhanced = pipe_enhanced.predict_proba(Xe_val)[:, 1]
accs_enhanced = [accuracy_score(ye_val, (p_val_enhanced >= t).astype(int)) for t in ths]
best_t_enhanced = float(ths[int(np.argmax(accs_enhanced))])


#Evaluate on TEST at those best thresholds
rows = []
rows.append(scores(pipe_base, Xb_test, y_test, 0.50,          "LR baseline (0.50)"))
rows.append(scores(pipe_enhanced,  Xe_test, y_test, 0.50,          "LR +LRI (0.50)"))
rows.append(scores(pipe_base, Xb_test, y_test, best_t_base,   f"LR baseline (best={best_t_base:.2f})"))
rows.append(scores(pipe_enhanced,  Xe_test, y_test, best_t_enhanced,    f"LR +LRI (best={best_t_enhanced:.2f})"))

performance_lr = pd.DataFrame(rows)
display(performance_lr)
```
